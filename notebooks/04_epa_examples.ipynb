# 04 · PM-EPA worked examples
Real possessions, their observed states, V values and credit rows (same as docs/pm_epa/attribution_examples.md Part B).

```python
import os, sys, json
sys.path.insert(0, os.path.abspath('..'))
import polars as pl, numpy as np
import matplotlib.pyplot as plt
from src.config import load_config
cfg = load_config()
proc = cfg.path(cfg.data.processed_dir)
pl.Config.set_tbl_rows(25); pl.Config.set_tbl_width_chars(160)
```

```python
tr = pl.read_parquet(proc / 'transitions.parquet')
tr.group_by('seg_type').agg(pl.len(), pl.col('delta').mean(), pl.col('gap').median()).sort('seg_type')
```

```python
cr = pl.read_parquet(proc / 'credits.parquet')
cr.group_by('component').agg(pl.len(), pl.col('value').sum()).sort('component')
```
